```python
from transformers import pipeline
```

```python
classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")
```

```python
categories=["Outdoor Gear","Kitchen Appliances","Apparel"]
```

```python
phrase1=" My new waterproof jacket and hiking boots are essential for my backpacking trip."
phrase2="The smart toaster and air fryer are the most-used appliances in our modern kitchen."
phrase3=" She chose a stylish dress and a comfortable pair of sneakers for her casual outing."
```

```python
predict=classifier(phrase2, candidate_labels=categories)
predict
```
